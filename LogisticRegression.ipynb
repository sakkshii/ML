```python
scaler= StandardScaler()
x_train= scaler.fit_transform(x_train)
x_test= scaler.transform(x_test)
```

```python
# start from sigmod
def sigmoid(z):
  return 1/ (1 + np.exp(-z))
```

```python
# test sigmiod
print(sigmoid(0))
print(sigmoid(100))
print(sigmoid(-100))
```

```python
# initialize weight / z parameters
def initialize_params(n):
  w= np.zeros(n)
  b= 0.0
  return w,b
```

```python
w,b= initialize_params(5)
print(w)
print(b)
```

```python
# forward pass
def forward(X, w, b):
  z= np.dot(X, w) + b
  y_hat= sigmoid(z)
  return y_hat
```

```python
# loss function
def loss(y, y_hat):
  m = len(y)
  loss = - np.mean(y*np.log(y_hat) + (1-y)*np.log(1-y_hat))
  return loss
```

```python
# backward pass, gradient descent
def backward(X, y, y_hat):
  m= len(y)
  dw = (1/m)*np.dot(X.T, (y_hat - y))
  db = (1/m)*np.sum(y_hat - y)
  return dw, db
```

```python
# train
def train(X, y, learning_rate=0.01, epochs=100):
  w,b = initialize_params(X.shape[1])

  for epoch in range(epochs):
    y_hat = forward(X, w, b)
    loss_value = loss(y, y_hat)
    dw,db = backward(X, y, y_hat)

    # update weights
    w= w - learning_rate * dw
    b = b- learning_rate * db

    if epoch%100 == 0:
      print(f"Epoch: {epoch}, Loss: {loss_value}")

  return w,b
```

```python
def predict(X, w, b):
  y_hat = forward(X, w, b)
  y_pred = np.where(y_hat > 0.5, 1, 0)
  return y_pred
```

```python
train(x_train, y_train)
```
